# Feature Engineering Pipeline

```
import pandas as pd
from src.cropyield_pipeline import AgriFeatureEngineer
df=pd.read_csv("../data/sample/crop_yield_demo.csv")
X=df.drop(columns="Yield_ton_ha")
engineered=AgriFeatureEngineer().fit_transform(X)
engineered[["NPK_Total","N_to_P_Ratio","K_to_N_Ratio","Rainfall_Category","Humidity_Temperature_Index","Growing_Degree_Days"]].head()
```

Feature rationale:
- NPK_Total: total of primary nutrients.
- N_to_P_Ratio and K_to_N_Ratio: relative nutrient balance.
- Rainfall_Category: interpretable rainfall bands.
- Humidity_Temperature_Index: humidity-temperature composite.
- Growing_Degree_Days: simple temperature accumulation proxy.

All feature transformations are inside the sklearn Pipeline to avoid preprocessing leakage.
